### Python Modules : 
* Modules are pre defined files that contain the python codes which imcludes functionality of class, methods and variables. etc.
* It consists of different functions , classes in a group of files inside the directory.
* Modules can also be termed as libraries.
* These are pre defined methods that can be used to make the code more efficient and reduce redundancy.
* A module contains are accessed with "import statement".
* A library is collection of modules, Many libraries only consists of a single module.

```python
#Histogram : In histogram , on x axis it is represent variables and on y axis it represents frequency of that variables.

#We have a sample data for blood sugar level of different


blood_sugar = [133,85,90,150,149,88,93,115,135,80,77,82,129]
plt.hist(blood_sugar , rwidth = 0.8)
```

```python
plt.hist(blood_sugar , rwidth = 0.8 , bins =4)
```

```python
# Historam showing normal, pre-diabetic and diabetic distribution:

# 80-100 => Normal
# 100-120 => pre-diabetic
# 120-150 →> Diabetic
```

```python
plt.xlabel("Sugar level")
plt.ylabel("Number of Patients") 
plt.title("Blood Sugar Chart")
plt.hist(blood_sugar, bins=[80,100,120,150], rwidth=0.9, color='g')
```

```python
plt.xlabel("Sugar Level")
plt.ylabel("Number of Patients") 
plt.title("Blood Sugar Chart")
plt.hist(blood_sugar, bins=[80,100,120,150], rwidth=0.7 , color='g',histtype='step')
```

```python
plt.xlabel("Sugar Level")
plt.ylabel("Number of Patients") 
plt.title("Blood Sugar ChART")
plt.hist(blood_sugar, width= 0.95, orientation='horizontal')
```

```python
# Pie Chart :Draw a pie chart to track down home expenses,
exp_val = [14000, 6000, 3000,2000, 5000]
exp_label = ['Home-Rent', 'Food','Groceries',' Other - bills', 'Savings']
plt.pie(exp_val,labels=exp_label)
```

```python
plt.pie(exp_val, labels=exp_label, shadow=True)
plt.axis("equal")
plt.show()
```

```python
plt.axis("equal")
plt.pie(exp_val, labels=exp_label, shadow=True, autopct='%1.2f%%', radius=1)
plt.show()
```

```python
plt.axis("equal")
plt.pie(exp_val, labels=exp_label, shadow= True, autopct= '%0.3f%%', radius= 1.5, explode=[0,0.2,0,0,0],
        counterclock= False, startangle=45)
plt.legend(loc='upper right')
plt.show()
```

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
import warnings 
warnings.filterwarnings("ignore")
```

```python
sns.load_dataset("tips")
```

```python
df = sns.load_dataset("tips")
df
```

```python
#Jointplot : A jointplot allows to study the relationship between two numeric variables.
```

```python
sns.jointplot(x= 'tip', y = 'total_bill' , data = df, kind = 'hex')
```

```python
sns.jointplot(x='tip', y= 'total_bill' , data = df , kind = 'reg')
```

```python
#Pair plot : It is also called sctter plot , in which one variable in the same data row is marked with another variable value.
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df, hue = 'sex')
```

```python
#Distplot : It is use to check distribution of the colomn features
```

```python
sns.distplot(df['tip'])
```

```python
#Categorial plot : Sea born helps us to do analysis on categorial data points for this we count and box plot.
```

```python
#Count Plot:
sns.countplot(y = 'sex' , data = df)
```
