#### Analysis with Python Project on Seaborne Library "IRIS"

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
import warnings 
warnings.filterwarnings("ignore")
```

```python
sns.load_dataset("iris")
```

```python
df = sns.load_dataset("iris")
df
```

```python
# This shows scatterplots, histograms, KDEs for all pairs
sns.pairplot(df, hue="species")
plt.suptitle("Pairplot of Iris Features", y=1.02)
plt.show()
```

```python
#Pairplot of particular category species
sns.pairplot(df, hue ='species') 
```

```python
#Dist plot for petal lenght
sns.distplot(df['petal_length']) 
```

```python
sns.countplot(y = 'species' , data = df) #Countplot
```

```python
#Scatter Plot
sns.scatterplot(data= df, x="sepal_length", y="sepal_width", hue="species")
plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.show()
```

```python
# Line Plot Visualization
iris_sorted = df.sort_values(by="sepal_length")

sns.lineplot(data=df, x="sepal_length", y="sepal_width", hue="species")
plt.title("Line Plot of Sepal Length vs Sepal Width")
plt.show()
```

```python
#Histogram
sns.histplot(data=df, x="petal_length", hue="species", multiple="stack")
plt.title("Histogram of Petal Length")
plt.show()
```

```python
#Box Plot
sns.boxplot(data=df, x="species", y="sepal_length")
plt.title("Box Plot of Sepal Length by Species")
plt.show()
```

```python
#Violin_plot
sns.violinplot(data=df, x="species", y="petal_length")
plt.title("Violin Plot of Petal Length by Species")
plt.show()
```

```python
#Srtip Plot
sns.stripplot(data=df, x="species", y="sepal_width")
plt.title("Strip Plot of Sepal Width by Species")
plt.show()
```

```python
#SwarmPlot
sns.swarmplot(data=df, x="species", y="petal_width")
plt.title("Swarm Plot of Petal Width by Species")
plt.show()
```

```python
# HeatMap Compute correlation matrix and drop non-numeric column
corr = df.drop(columns="species").corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Heatmap of Feature Correlation")
plt.show()
```
